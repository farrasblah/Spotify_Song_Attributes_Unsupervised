```python
# Handling outlier menggunakan IQR dan Mean +- 3 sd

# from sklearn.preprocessing import RobustScaler

# scaler = RobustScaler()
# pca_df_scaled = pd.DataFrame(scaler.fit_transform(pca_df), columns=pca_df.columns)

# Fungsi deteksi outlier (Mean ± 3*SD) longgarkan batasnya menjadi 1.8 IQR atau 3.5 SD
def detect_outliers_mean_sd(series, threshold=3.5):
    mean = series.mean()
    std = series.std()
    lower_bound = mean - threshold * std
    upper_bound = mean + threshold * std
    return (series < lower_bound) | (series > upper_bound)

# Fungsi deteksi outlier (IQR)
def detect_outliers_iqr(series, factor=1.8):
    Q1 = series.quantile(0.25)
    Q3 = series.quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - factor * IQR
    upper_bound = Q3 + factor * IQR
    return (series < lower_bound) | (series > upper_bound)

method_I = ['PCA_3', 'PCA_5', 'PCA_6', 'PCA_7']
method_II = ['PCA_1', 'PCA_2', 'PCA_4']

# Mendeteksi outlier di pca_df
outliers = pd.DataFrame()

for col in pca_df.columns:
    if col in method_I:
        outliers[col] = detect_outliers_mean_sd(pca_df[col], threshold=4)
    elif col in method_II:
        outliers[col] = detect_outliers_iqr(pca_df[col], factor=2)

print("Jumlah outlier di setiap kolom:")
print(outliers.sum())

cleaned_pca_df = pca_df[~outliers.any(axis=1)]

print(f"Data sebelum pembersihan: {pca_df.shape[0]} baris")
print(f"Data setelah pembersihan: {cleaned_pca_df.shape[0]} baris")

plt.figure(figsize=(16, 10))
for i, col in enumerate(pca_df.columns, 1):
    plt.subplot(3, 3, i)
    sns.histplot(cleaned_pca_df[col], kde=True, bins=40, color=sns.color_palette("Set2")[i - 1])
    plt.title(f'Distribusi {col} (Setelah Pembersihan)')

plt.tight_layout()
plt.show()
```

```python
# K means tp nyari k terbaik dlu

from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

wcss = []
scores = []
for i in range(2, 10):
    km = KMeans(n_clusters=i, init='k-means++', max_iter=300, n_init=10, random_state=42)
    km.fit(cleaned_pca_df)
    wcss.append(km.inertia_)

    labels = km.labels_
    silhouette_avg = silhouette_score(cleaned_pca_df, labels)
    scores.append(silhouette_avg)
    print('wcss score for n_cluster = ' + str(i) + ' is ' + str(wcss))
    print('silhoutte score for n_clusters = ' + str(i) + ' is ' + str(silhouette_avg))
```

```python
#  Visualisasi Elbow Method
plt.plot(range(2, 10), wcss)
plt.title('The Elbow Method', fontsize = 20)
plt.xlabel('No. of Clusters')
plt.ylabel('WCSS')
plt.show()
```

```python
# Visualisasi Silhouette Score
plt.plot(range(2, 10), scores)
plt.title('Silhouette Score', fontsize = 20)
plt.xlabel('No. of Clusters')
plt.ylabel('Score')
plt.show()
```

```python
# Hierarchical clustering

from sklearn.cluster import AgglomerativeClustering

linkage_col = ['ward', 'complete', 'average', 'single']
scores_all = [[] for _ in range(len(linkage_col))]
for j in range (len(linkage_col)):
    print('Linkage: ', linkage_col[j])
    scores = []
    for i in range(2, 10):
        AC = AgglomerativeClustering(n_clusters=i, linkage = linkage_col[j]) 
        AC.fit(cleaned_pca_df)

        labels = AC.labels_
        silhouette_avg = silhouette_score(cleaned_pca_df, labels)
        scores.append(silhouette_avg)
        print('silhoutte score for n_clusters = ' + str(i) + ' is ' + str(silhouette_avg))
    print("\n##########################\n")
    scores_all[j] = scores
```

```python
colors = sns.color_palette("Set1", len(linkage_col))

for j in range(len(linkage_col)):
    plt.plot(range(2, 10), scores_all[j], marker='o', color=colors[j], label=linkage_col[j])

plt.xlabel('Number of Clusters')
plt.ylabel('Silhouette Score')
plt.title('Silhouette Score vs. Number of Clusters for Different Linkage Methods')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import plotly.graph_objects as go
from sklearn.cluster import AgglomerativeClustering
from scipy.cluster.hierarchy import dendrogram, linkage

banyak_cluster = 4

AC = AgglomerativeClustering(n_clusters=banyak_cluster, linkage='ward')
labels = AC.fit_predict(cleaned_pca_df)

pca_with_labels = np.column_stack((cleaned_pca_df.values, labels))

centroids = []
for cluster_label in range(banyak_cluster):
    cluster_data = pca_with_labels[pca_with_labels[:, -1] == cluster_label]
    cluster_centroid = np.mean(cluster_data[:, :-1], axis=0)
    centroids.append(cluster_centroid)
centroids = np.array(centroids)

traces = []
colors = ['red', 'blue', 'green', 'yellow']

for cluster_label in range(banyak_cluster):
    cluster_data = pca_with_labels[pca_with_labels[:, -1] == cluster_label]
    trace = go.Scatter3d(
        x=cluster_data[:, 0],  # Sumbu X
        y=cluster_data[:, 1],  # Sumbu Y
        z=cluster_data[:, 2],  # Sumbu Z
        mode='markers',
        marker=dict(size=8, color=colors[cluster_label], opacity=0.8),
        name=f'Cluster {cluster_label + 1}'
    )
    traces.append(trace)

centroid_trace = go.Scatter3d(
    x=centroids[:, 0],
    y=centroids[:, 1],
    z=centroids[:, 2],
    mode='markers',
    marker=dict(size=12, color='black', symbol='diamond', opacity=1),
    name='Centroids'
)
traces.append(centroid_trace)

layout = go.Layout(
    title='Hasil Agglomerative Clustering pada cleaned_pca_df',
    scene=dict(
        xaxis_title='PC1',
        yaxis_title='PC2',
        zaxis_title='PC3'
    ),
    showlegend=True
)

fig = go.Figure(data=traces, layout=layout)
fig.show()
```

```python
# pake DBSCAN (pakai data sebelum di handle outliernya karena DBSCAN bisa digunakan untuk menghandle outlie)

from sklearn.cluster import DBSCAN
from sklearn.metrics import silhouette_score

df_train = pca_df.copy()

epsilon = np.arange(0.1, 6, 0.1) 
min_samples = range(1, 26)        

sil_avg = []
max_value = (0, 0, 0, -1)  

for eps in epsilon:
    for min_s in min_samples:
        db = DBSCAN(eps=eps, min_samples=min_s).fit(df_train)

    
        labels = db.labels_
        unique_labels = set(labels)

        # Periksa apakah ada setidaknya dua cluster
        if len(unique_labels) >= 2:
            # Hitung Silhouette Score
            silhouette_avg = silhouette_score(df_train, labels)
            sil_avg.append(silhouette_avg)

            if silhouette_avg > max_value[3]:
                max_value = (eps, min_s, len(unique_labels) - (1 if -1 in labels else 0), silhouette_avg)

print(f"Epsilon = {max_value[0]:.2f}")
print(f"min_samples = {max_value[1]}")
print(f"Jumlah cluster = {max_value[2]}")
print(f"Silhouette score = {max_value[3]:.4f}")
```

```python
# hasilnya cukup aneh, kurang tahu kenapa karena saya hanya mencoba

best_epsilon = max_value[0]
best_min_samples = max_value[1]

db = DBSCAN(eps=best_epsilon, min_samples=best_min_samples, metric='euclidean')
db.fit(df_train)
best_labels = db.labels_

unique_labels = np.unique(best_labels)
print(f"Jumlah Cluster (termasuk noise) = {len(unique_labels)}")

# Warna untuk tiap cluster (label -1 adalah outlier)
colors = ['blue', 'red', 'green', 'purple', 'orange', 'brown', 'pink', 'gray', 'yellow']
cluster_colors = [colors[label % len(colors)] if label != -1 else 'black' for label in best_labels]

trace = go.Scatter3d(
    x=df_train.iloc[:, 0],  # PCA_1
    y=df_train.iloc[:, 1],  # PCA_2
    z=df_train.iloc[:, 2],  # PCA_3
    mode='markers',
    marker=dict(size=8, color=cluster_colors, opacity=0.8),
    text=[f'Cluster {label}' if label != -1 else 'Outlier' for label in best_labels],
    name='Data Points'
)

# Layout grafis
layout = go.Layout(
    title=f'Hasil DBSCAN Clustering (epsilon={best_epsilon}, min_samples={best_min_samples})',
    scene=dict(
        xaxis_title='PCA 1',
        yaxis_title='PCA 2',
        zaxis_title='PCA 3'
    ),
    showlegend=False
)

fig = go.Figure(data=[trace], layout=layout)

fig.show()
```
